# Day 01 — Excel → GIS Update Automation

## Problem
A GIS team receives recurring Excel records that must be matched to existing GIS features. The workflow must validate the incoming data before any GIS edits occur, update only matched assets, and report invalid or unmatched records.

## Developer principle
**Validate first. Edit second.**

The goal is to separate input validation from GIS editing so bad input can never partially modify the production dataset.

## Expected Excel schema

The `updates` sheet must contain:

- `asset_id` — join key
- `status_new` — new GIS status
- `priority_new` — new GIS priority
- `inspector_new` — new inspector
- `inspection_date` — new inspection date

`AST-999` is intentionally absent from the GIS dataset and should be reported as unmatched rather than causing the process to fail.
